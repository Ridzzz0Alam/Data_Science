```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
```

# Step 1: Read the dataset

```python
df = pd.read_csv('income.csv')
print("Original Dataset:")
print(df)
print("\nShape:", df.shape)
```

# Step 2: Identify numerical features to scale
'Name' is a string identifier, 'Buy' is a categorical target    

 Only 'Age' and 'Income' are numerical features relevant for scaling

```python
numerical_features = ['Age', 'Income']

print("Numerical features selected for scaling:", numerical_features)
print("\nStatistics before scaling:")
print(df[numerical_features].describe())
```

# Step 3: Apply Standardization (Z-score scaling)

Formula: z = (x - mean) / std

Result: mean = 0, std = 1 for each feature


```python
scaler = StandardScaler()
```

# Fit and transform the numerical features


```python
df_scaled = df.copy()
df_scaled[numerical_features] = scaler.fit_transform(df[numerical_features])

print("Dataset after Standardization:")
print(df_scaled)
```

# Step 4: Verify the scaling results

```python
print("Statistics after Standardization:")
print(df_scaled[numerical_features].describe())

print("\nMean of scaled features (should be ~0):")
print(df_scaled[numerical_features].mean().round(6))

print("\nStd of scaled features (should be ~1):")
print(df_scaled[numerical_features].std().round(6))
```

# Step 5: Display the scaler parameters learned from data

```python
print("Scaler Mean (used for standardization):")
for feature, mean in zip(numerical_features, scaler.mean_):
    print(f"  {feature}: {mean:.4f}")

print("\nScaler Std (used for standardization):")
for feature, std in zip(numerical_features, scaler.scale_):
    print(f"  {feature}: {std:.4f}")
```
