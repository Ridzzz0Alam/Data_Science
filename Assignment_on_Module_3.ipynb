**Q1. Create a Python class representing a student with attributes such as name and ID, and include both a default constructor and a parameterized constructor. Add a method to display the student details, and use the pass statement inside an empty placeholder method. Then create multiple objects from this class to test each constructor.**

```python
class Student:
    total_students = 0
    def __init__(self):
        # Default constructor- initializes with default values
        self.name="Unknown"
        self.student_id=0
        Student.total_students+=1

    # Parameterized Constructor
    def __init__(self,name,student_id):
        # Parameterized constructor- initializes with provided values
        self.name=name
        self.student_id=student_id
        Student.total_students+=1

    def display_details(self):
        # Method to display student details
        print(f"Name:{self.name}")
        print(f"Student ID:{self.student_id}")
        print("-"*40)

    def placeholder_method(self):
        # Empty placeholder method using pass statement
        pass

    @classmethod
    def get_total_students(cls):
        # Class method to get total students created
        return cls.total_students

# Testing the Student class
print("=" * 40)
print("Q1: CONSTRUCTORS AND OBJECT CREATION")
print("=" * 40)

# Create objects using parameterized constructor
student1 = Student("Thibaut Courtois", 101)
student2 = Student("Eduardo Camavinga", 102)
student3 = Student("Amad Diallo", 103)

print("\nStudent Details:")
student1.display_details()
student2.display_details()
student3.display_details()

print(f"Total Students Created: {Student.get_total_students()}\n")
```

**Q2. Write a Python program demonstrating single, multilevel, and hierarchical inheritance using simple classes such as Person, Student, and Teacher. Include at least one overridden method in the child classes to show method overriding.**

```python
# Base Class
class Person:
    def __init__(self, name, age):
        self.name=name
        self.age=age

    def display_info(self):
        return f"Name:{self.name},Age:{self.age}"

# Single Inheritance- Student inherits from Person
class Student_Q2(Person):
    def __init__(self,name,age,student_id,grade):
        super().__init__(name, age)
        self.student_id=student_id
        self.grade=grade

    # Method Overriding
    def display_info(self):

        #Overridden method with additional student info
        parent_info=super().display_info()
        return f"{parent_info},Student ID:{self.student_id},Grade:{self.grade}"

# Multilevel Inheritance- GraduateStudent inherits from Student
class GraduateStudent(Student_Q2):

    #GraduateStudent class- multilevel inheritance
    def __init__(self,name,age,student_id,grade,thesis_topic):
        super().__init__(name,age,student_id,grade)
        self.thesis_topic=thesis_topic

    # Method Overriding
    def display_info(self):
        #Overridden method with thesis information
        parent_info=super().display_info()
        return f"{parent_info},Thesis:{self.thesis_topic}"

# Hierarchical Inheritance- Teacher also inherits from Person
class Teacher(Person):

    #Teacher class- hierarchical inheritance from Person
    def __init__(self, name, age, employee_id, subject):
        super().__init__(name,age)
        self.employee_id=employee_id
        self.subject=subject

    # Method Overriding
    def display_info(self):
        """Overridden method with teacher info."""
        parent_info=super().display_info()
        return f"{parent_info},Employee ID:{self.employee_id},Subject:{self.subject}"

# Testing Inheritance
print("\n---Single Inheritance (Student)---")
student=Student_Q2("Bruno Fernandes",20,201,"A")
print(student.display_info())

print("\n---Multilevel Inheritance (GraduateStudent)---")
grad_student=GraduateStudent("Jude Bellingham",25,301,"A+","AI in Healthcare")
print(grad_student.display_info())

print("\n---Hierarchical Inheritance (Teacher)---")
teacher=Teacher("Kylian Mbappe",40,501,"Mathematics")
print(teacher.display_info())

print()
```

**Q3. Define a class that shows encapsulation by using private attributes and public getter/setter methods. Add two methods with the same name but different parameter counts to illustrate method overloading (using default arguments). Then create another class to demonstrate multiple inheritance.**

```python
print("\n" + "=" * 40)
print("Q")

class BankAccount:
  def __init__(self, account_holder, balance=0):
    self.account_holder=account_holder
    self.balance=balance
    self.account_number=self.generate_account_number()

  def generate_account_number(self):
    import random
    return random.randint(100000,999999)

  # Getter Methods
  def get_account_holder(self):
    return self.account_holder

  def get_balance(self):
    return self.balance

  def get_account_number(self):
    return self.account_number

  # Setter Methods
  def set_account_holder(self,name):
    if name and isinstance(name,str):
      self.account_holder=name
    else:
      print("Invalid name!")

  def set_balance(self,amount):
    if amount>=0:
      self.balance=amount
    else:
      print("Balance cannot be negative")

# METHOD OVERLOADING

  def deposit(self,amount,bonus=0,description="Regular deposit"):
    total=amount+bonus
    self.balance+=total
    print(f"Transaction: {description}")
    print(f"Deposited: {amount:.2f}")
    if bonus>0:
      print(f"Bonus: ${bonus:.2f}")
    print(f"Total Added: ${total:.2f}")
    print(f"New Balance: ${self.balance:.2f}")

  def withdraw(self,amount,fee=0):
    total=amount+fee
    if self.balance>=total:
      self.balance-=total
      print(f"Withdrawn: ${amount:.2f}")
      if fee>0:
        print(f"Fee: ${fee:.2f}")
      print(f"Total Deducted: ${total:.2f}")
      print(f"New Balance: ${self.balance: 2f}")
    else:
      print(f"Insufficient balance: ${self.balance:.2f}")

  def display_account_info(self):
    print(f"Account Number: {self.account_number}")
    print(f"Account Holder: {self.account_holder}")
    print(f"Balance: ${self.balance:.2f}")


# MULTIPLE INHERITANCE

class Printable:

  def print_details(self):
    print("\n" + "=" *50)
    print("DETAILED INFORMATION")
    print("=" * 50)

class Saveable:

  def save_to_file(self,filename):
    print(f"\nSaving data to {filename}...")
    print("Data saved successfully!")

class PremiumBankAccount(BankAccount,Printable,Saveable):

  def __init__(self,account_holder,initial_balance=0,interest_rate=0.05):
    super().__init__(account_holder,initial_balance)
    self.interest_rate=interest_rate

  def apply_interest(self):
    interest=self.get_balance()*self.interest_rate
    self.deposit(interest,description="Interest credit")

  def print_details(self):
    super().print_details()
    self.display_account_info()
    print(f"Interest Rate: {self.interest_rate*100}%")
    print(f"Account Type: Premium")
    print("="*50)

# TESTING

print("DEMONSTRAITNG ENCAPSULATION")

account1=BankAccount("Nazmul Hassan", 1000)
account1.display_account_info()

print("\n---TESTING GETTERS AND SETTERS---")
print(f"Account Holder: {account1.get_account_holder()}")
print(f"Balance: ${account1.get_balance():.2f}")
print(f"Account Number: {account1.get_account_number()}")
account1.set_account_holder("Nazmul Hassan Shanto")
print(f"Updated Name: {account1.get_account_holder()}")


# Method overlaoding with different parameter counts

print("\n---Deposit with 1 parameter---")
account1.deposit(500)

print("\n---Deposit with 2 parameters---")
account1.deposit(300,50)

print("\n---Deposit with 3 parameters---")
account1.deposit(200,25, "Promotional deposit")

print("\n---Withdraw with 1 parameter---")
account1.withdraw(100)

print("\n---Withdraw with 2 parameters---")
account1.withdraw(200,5)

print("MULTIPLE INHERITANCE")

# Create premium account with multiple inheritance
premium_account = PremiumBankAccount("Ahmed Shah",5000,0.08)

print("\n---Using methods from BankAccount---")
premium_account.display_account_info()

print("\n---Using methods from Printable---")
premium_account.print_details()

print("\n---Using methods from Saveable---")
premium_account.save_to_file("premium_account.dat")

print("\n---Premium Account Features---")
premium_account.apply_interest()

print("\n---Method Resolution Order (MRO)---")
print("MRO:",[cls.__name__ for cls in PremiumBankAccount.__mro__])
```
