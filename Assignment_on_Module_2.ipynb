# **1) Write a Python program that takes a list of numbers as input, removes duplicates using a suitable list method, and returns a dictionary containing the original list, the unique values, and their count. Use a function with parameters and a return statement to perform this task.**

```python
def list_unique_summary(numbers):

    original_list=numbers[:]
    unique_list=[]

    for n in numbers:
        if n not in unique_list:
            unique_list.append(n)

    return{"original_list": original_list, "unique_values": unique_list,"unique_count": len(unique_list)}

nums=[1,2,2,3,4,4,5,1]
result=list_unique_summary(nums)
print(result)
```

# **2) Create a function that accepts two sets as parameters and returns their union, intersection, and difference. Use keyword arguments with default parameter values so the function can work even if one of the sets is not provided by the user. Display the results clearly.**

```python
def set_operations(set1=None, set2=None):

    if set1 is None:
        set1=set()
    if set2 is None:
        set2=set()

    return{"union": set1 | set2,"intersection": set1 & set2,"difference_set1_minus_set2": set1 - set2}

A={1,2,3,4}
B={3,4,5,6}

ops=set_operations(set1=A,set2=B)
print("Union:",ops["union"])
print("Intersection:",ops["intersection"])
print("Difference (set1 - set2):",ops["difference_set1_minus_set2"])

ops2=set_operations(set1=A)
print("\nOnly set1 provided:")
print("Union:",ops2["union"])
print("Intersection:",ops2["intersection"])
print("Difference (set1 - set2):",ops2["difference_set1_minus_set2"])
```
